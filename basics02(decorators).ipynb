## 파이썬 기초 복습(2편: decorator)

파이썬 데코레이터(decorator)는
  * 기존의 함수나 클래스를 수정하지 않고
  * 기능을 추가하거나 변경할 수 있게 해주는 도구입니다.

1. 함수는 객체(특히 일급객체)다.
  * 변수에 할당하거나, 다른 함수의 인자로 넘기거나, 함수에서 반환할 수 있습니다.

```python
def greet():
    return "안녕"

say = greet
print(say())
```

2. 함수를 인자로 전달해봅시다

```python
def hello():
    print('안녕')

# 함수를 인자로 받아서 2번 실행하는 함수
def run_twice(func):
    func()
    func()

run_twice(hello)
```

3. 함수 안에서 새로운 함수를 만들어 리턴해 보겠습니다.
  * 이게 데코레이터의 진짜 핵심입니다.

```python
def make_inner():
    def inner():    # make_inner 내에서만 정의됨(외부에서 접근X)
        print('[inner] 안녕~')
    return inner    # inner 함수 자체를 돌려줌 (실행 X)

# inner()           # 에러

greeter = make_inner()  # greeter에는 inner 함수가 담김.
greeter()


# make_inner()()    # 이렇게 해도 실행됨
```

make_greeter()를 호출하면 그 안에서 inner라는 함수를 새로 정의하고, 그것을 그대로 반환합니다. 그래서 greeter는 사실상 inner 함수가 됩니다.

4. 받은 함수를 "포장"해서 돌려주기
  * 이제 2단계와 3단계를 합칩니다.
  * 함수를 인자로 받아서 → 그것을 감싼 새 함수를 내부에서 만들어 → 리턴하는 함수를 만듭니다. ==> 이게 바로 데코레이터입니다.
  * 쉬운 비유로, 선물(원래 함수)을 받아서 포장지(추가 기능)로 감싸 돌려주는 것과 같습니다.

```python
def decorator(func):           # 원래 함수를 선물로 받음
    def wrapper():             # 포장지 역할을 할 새 함수(이게 리턴됨)
        print("=" * 20)        # 포장: 위쪽 장식
        func()                 # 원래 선물(함수)을 그대로 실행
        print("=" * 20)        # 포장: 아래쪽 장식
    return wrapper             # 포장 완료된 것을 돌려줌

def say_hello():
    print("Hello!")

new_hello = decorator(say_hello)   # say_hello를 포장하기!
new_hello()
```

```python
# 실무적으로는 그냥 이름을 덮어쓴다.
say_hello = decorator(say_hello)
say_hello()
```

```python
# 퀴즈: 이미 덮어쓰기 당한 say_hello를 다시 데코레이터로 포장하면 결과는 어떨까?
say_hello = decorator(say_hello)
say_hello()
```

5. @를 씌워보자.

```python
def decorator(func):           # 원래 함수를 선물로 받음
    def wrapper():             # 포장지 역할을 할 새 함수(이게 리턴됨)
        print("=" * 20)        # 포장: 위쪽 장식
        func()                 # 원래 선물(함수)을 그대로 실행
        print("=" * 20)        # 포장: 아래쪽 장식
    return wrapper             # 포장 완료된 것을 돌려줌

def say_hello():
    print("Hello!")
```

say_hello = decorator(say_hello)

이 패턴("함수를 정의한 뒤, 같은 이름으로 데코레이터에 통과시켜 다시 할당")이 너무 자주 쓰여서,

파이썬이 짧게 쓰는 문법을 만들어 줬습니다. 그게 @입니다.

아래 두 코드는 완전히 동일하게 동작합니다.




```python
# 방법 A: 직접 쓰기
def say_hello():
    print("Hello!")

say_hello = decorator(say_hello)
```

```python
# 방법 B: @ 사용 (방법 A와 100% 동일)
@decorator
def say_hello():
    print("Hello!")
```

@decorator를 함수 정의 바로 위에 붙이면, 파이썬은 속으로 이렇게 해석합니다.

  - "이 say_hello 함수를 정의한 직후에, say_hello = decorator(say_hello)를 자동으로 실행해 줄게."
  - 당연히 say_hello라는 이름은 decorator를 통과한 결과값(함수)을 갖게 되므로, def의 내용과는 달라진다.
