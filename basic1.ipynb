## 파이썬 기초 복습

1. packing과 unpacking

  * 파이썬에서 packing과 unpacking은 함수 인자를 다룰 때 매우 강력한 기능.
  
  * 똑같은 *과 ** 연산자가, 어디에서 사용하느냐(위치하느냐)에 따라 의미가 달라진다는 점.

  * <핵심> 함수 정의부에서 사용하면 packing, 함수 호출부에서 사용하면 unpacking이 된다.


  + Packing (포장하기) : "함수 정의" 부분에서 사용
  
    + "*" : 여러 개의 positinal parameter가 들어간 함수call을, 함수 측에서 단 하나의 Tupule 변수로 "묶어서" 받는다.

```python
# *args — 위치 인자 packing

def sum_all(*args):
    print(type(args))  # <class 'tuple'>
    print(args)
    return

sum_all(1, 2, 3)        # args = (1, 2, 3)
sum_all(1, 2, 3, 4, 5)  # args = (1, 2, 3, 4, 5)
```

```python
# 주의 : sum_all(1,2,3)과 sum_all((1,2,3))은 다르다.
sum_all(1,2,3)
sum_all((1,2,3))

# 당연히 이것과도 다르다
sum_all((1,2),3)
```

  + Packing (포장하기) : "함수 정의" 부분에서 사용
    + "**" : 여러 개의 named parameter가 들어간 함수call을, 함수 측에서 단 하나의 Dict 변수로 "묶어서" 받는다.

```python
# **kwargs — 키워드 인자 packing

def print_info(**kwargs):
    print(type(kwargs))  # <class 'dict'>
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_info(name="Kim", age=30, job="developer")
# kwargs = {'name': 'Kim', 'age': 30, 'job': 'developer'}
```

```python
# 두 개를 섞어서 쓰기(packing 총정리)

def flexible(a, b, *args, **kwargs):
    print(f"a={a}, b={b}")
    print(f"args={args}")
    print(f"kwargs={kwargs}")

flexible(1, 2, 3, 4, 5, name="Kim", age=30)
# a=1, b=2
# args=(3, 4, 5)
# kwargs={'name': 'Kim', 'age': 30}
```

주의: 순서는 반드시 다음 순으로 정의해야 합니다.

일반(positinal) 인자 → *args → 기본값(default) 인자 → **kwargs

2. Unpacking (풀어내기) — "함수 호출" 부분에서 사용


  * 이미 묶여 있는 컨테이너(list, tuple, dict 등)를 개별 인자로 풀어서 함수에 전달.

```python
# * : iterable unpacking

def greet(name, age, job):
    print(f"{name} ({age}) is a {job}")

person = {"name": "Kim", "age": 30, "job": "developer"}

greet(**person)  # greet(name="Kim", age=30, job="developer")와 동일
```

```python
# 만약 순서가 바뀐다면?

person2 = {"name": "Kim", "job": "developer", "age": 30}

greet(**person2)  # 순서 바뀌어도 상관없다
```

dict 앞에 **를 붙이면 key-value 쌍이 키워드 인자로 풀어집니다.

이때 dict의 key 이름이 함수의 매개변수 이름과 정확히 일치해야 합니다.

```python
# unpacking 종합(2개 동시에 사용하는 패턴)
# 이 둘은 함께 쓰일 때 진가를 발휘합니다. 특히 데코레이터나 함수 래핑 시에 자주 등장합니다.
# wrapper는 어떤 시그니처의 함수든 감쌀 수 있게 됩니다. 이게 *args, **kwargs의 진짜 위력이에요.

def logger(func):
    def wrapper(*args, **kwargs):  # packing: 어떤 인자든 받음
        print(f"호출: {func.__name__}({args}, {kwargs})")
        result = func(*args, **kwargs)  # unpacking: 원래 함수에 그대로 전달
        print(f"결과: {result}")
        return result
    return wrapper

@logger
def multiply(a, b, c=1):
    return a * b * c

multiply(2, 3, c=4)
# 호출: multiply((2, 3), {'c': 4})
# 결과: 24
```

3. 자주 헷갈리는 패턴

```python
# 1. 똑같은 *인데 의미는 정반대

def f(*args): ...   # 정의부의 * → packing (묶기)
f(*[1, 2, 3])       # 호출부의 * → unpacking (풀기)
```

```python
# 2. keyword 강제: 함수 정의에서 * 뒤에 오는 인자는 반드시 키워드로만 전달해야 합니다.

def f(a, *, b, c):
    return a, b, c

f(1, b=2, c=3)  # OK
f(1, 2, 3)      # TypeError!
```

```python
# 3. unpacking을 통한 병합
# 함수 호출뿐 아니라 컨테이너 리터럴 안에서도 unpacking이 가능합니다.


list1 = [1, 2]
list2 = [3, 4]
merged = [*list1, *list2, 5]      # [1, 2, 3, 4, 5]
print(merged)

dict1 = {"a": 1}
dict2 = {"b": 2}
merged = {**dict1, **dict2, "c": 3}  # {'a': 1, 'b': 2, 'c': 3}
print(merged)
```
