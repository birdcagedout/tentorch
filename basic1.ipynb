## 파이썬 기초 복습

1. packing과 unpacking

파이썬에서 packing과 unpacking은 함수 인자를 다룰 때 매우 강력한 기능입니다. 핵심은 *과 ** 연산자가 어디에 위치하느냐에 따라 의미가 달라진다는 점입니다.


  + Packing (포장하기) — "함수 정의"에서 사용
  
    여러 인자를 하나의 변수로 "묶어서" 받는 것을 의미합니다.

```python
def sum_all(*args):
    print(type(args))  # <class 'tuple'>
    print(args)
    return

sum_all(1, 2, 3)        # args = (1, 2, 3)
sum_all(1, 2, 3, 4, 5)  # args = (1, 2, 3, 4, 5)
```

```python
# 주의 : sum_all(1,2,3)과 sum_all((1,2,3))은 다르다.
sum_all(1,2,3)
sum_all((1,2,3))

# 당연히 이것과도 다르다
sum_all((1,2),3)
```
